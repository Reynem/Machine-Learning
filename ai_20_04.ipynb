```python
import requests, bs4
headers = {"User-Agent": "Mozilla/5.0"}
res = requests.get("https://en.wikipedia.org/wiki/Machine_learning", headers=headers)
soup = bs4.BeautifulSoup(res.text, 'html.parser')
x = soup.find_all('p')
text =''
for i in range(len(x)):
  text += x[i].text.strip() + '\n'
print(text)
```
